```python
import pandas as pd

df = pd.read_csv(r'D:\TOPS\PYTHON\PYTHON ASSESSMENT\SECTION B\sales_data.csv', encoding='latin1')

df['PRODUCTLINE'] = df['PRODUCTLINE'].astype(str).str.strip().str.title()

print("Data loaded and product names cleaned.")
```

```python
#TASK : Convert SALES to numbers cleanly

df['SALES'] = pd.to_numeric(df['SALES'], errors='coerce')

#TASK : Find high revenue outliers (anything over 5,000)

outliers = df[df['SALES'] > 5000]

print("SALES converted to float.")
print("Number of high-value outliers found:", len(outliers))
```

```python
#TASK : Calculate total sales for each product line

summary = df.groupby('PRODUCTLINE')['SALES'].sum().reset_index()

#TASK :  Categorize as High Revenue if total sales > 100,000

summary['Category'] = 'Standard Revenue'
summary.loc[summary['SALES'] > 100000, 'Category'] = 'High Revenue Product'

#TASK : Print the simple report

print(summary)
```
