```python
import pandas as pd

file_path = r'D:\TOPS\PYTHON\PYTHON ASSESSMENT\SECTION B\sales_data.csv'

df = pd.read_csv(file_path, encoding='latin1')

df['PRODUCTLINE'] = df['PRODUCTLINE'].astype(str).str.strip().str.title()

print("Cleaned PRODUCTLINE values:")
print(df['PRODUCTLINE'].head())
```

```python
#TASK : 2 : Iterate through the 'SALES' column to remove non-numeric characters and cast the cleaned values into a float data type.

df['SALES'] = pd.to_numeric(df['SALES'], errors='coerce')

print("SALES data type:", df['SALES'].dtype)
```

```python
#TASK : 3 : Construct dictionary mapping unique 'PRODUCTLINE' to cumulative sum of 'SALES'

product_sales_dict = {}

for prod, sales in zip(df['PRODUCTLINE'], df['SALES_CLEAN']):
    product_sales_dict[prod] = product_sales_dict.get(prod, 0.0) + sales

print("Dictionary Result:")
print(product_sales_dict)
```

```python
#TASK : 4 : Loop and conditional logic to categorize product lines exceeding 100,000

for product, total_sales in product_sales_dict.items():
    if total_sales > 100000:
        print(product, ": High Revenue Product")
    else:
        print(product, ": Standard Revenue Product")
```
